# GLOBAL RULES - APPLY TO ALL FILES

| Rule | Why |
|---|---|
| **UUIDs** → always `string` | Never convert to number, it breaks! |
| **ZIP / NPI** → always `string` | Leading zeros vanish if read as number |
| **Dates/Times** → ISO-8601 UTC | `YYYY-MM-DD` or `YYYY-MM-DDTHH:MM:SSZ` - NO local time |
| **PII Columns** → REMOVE/HASH before reporting | `SSN, FIRST, LAST, DRIVERS, PASSPORT` - NEVER reach reporting layer |

---

# REFERENTIAL INTEGRITY SUMMARY

> **Every encounter(appointments) MUST link to valid IDs in these 4 files:**
> - `PATIENT` → exists in `patients.Id`
> - `ORGANIZATION` → exists in `organizations.Id`
> - `PROVIDER` → exists in `providers.Id`
> - `PAYER` → exists in `payers.Id`
>
> **Every condition MUST link to:**
> - `ENCOUNTER` → exists in `encounters.Id`
