# GLOBAL RULES - APPLY TO ALL FILES

| Rule | Why |
|---|---|
| **UUIDs** → always `string` | Never convert to number — it breaks! |
| **ZIP / NPI** → always `string` | Leading zeros vanish if read as number |
| **Dates/Times** → ISO-8601 UTC | `YYYY-MM-DD` or `YYYY-MM-DDTHH:MM:SSZ` - NO local time |
| **PII Columns** → REMOVE/HASH before reporting | `SSN, FIRST, LAST, DRIVERS, PASSPORT` - NEVER reach reporting layer |

---

# REFERENTIAL INTEGRITY SUMMARY

> **Every encounter(appointments) MUST link to valid IDs in these 4 files:**
> - `PATIENT` → exists in `patients.Id`
> - `ORGANIZATION` → exists in `organizations.Id`
> - `PROVIDER` → exists in `providers.Id`
> - `PAYER` → exists in `payers.Id`
>
> **Every condition MUST link to:**
> - `ENCOUNTER` → exists in `encounters.Id`

This Data Dictionary is now the **SOURCE OF TRUTH** for every validation rule!

### Reading Data into pandas dataframe

```python
import pandas as pd
from pathlib import Path

# Define the path to the landing folder where the CSV files are stored
RAW_DATA_PATH = Path(r"C:\Users\dell\Desktop\caresync_project\caresync_project\data\landing")

# Load the CSV files into pandas DataFrames
patients = pd.read_csv(RAW_DATA_PATH / "patients.csv", low_memory=False)
encounters = pd.read_csv(RAW_DATA_PATH / "encounters.csv", low_memory=False)
organizations = pd.read_csv(RAW_DATA_PATH / "organizations.csv", low_memory=False)
payers = pd.read_csv(RAW_DATA_PATH / "payers.csv", low_memory=False)
providers = pd.read_csv(RAW_DATA_PATH / "providers.csv", low_memory=False)
conditions = pd.read_csv(RAW_DATA_PATH / "conditions.csv", low_memory=False)
```

###  Pandas Reading Template

```python
patients = pd.read_csv(RAW_DATA_PATH / "patients.csv",
    dtype={"Id": str, "SSN": str, "ZIP": str, "DRIVERS": str, "PASSPORT": str},
    parse_dates=["BIRTHDATE", "DEATHDATE"])

encounters = pd.read_csv(RAW_DATA_PATH / "encounters.csv",
    dtype={"Id": str, "PATIENT": str, "ORGANIZATION": str, "PROVIDER": str, "PAYER": str},
    parse_dates=["START", "STOP"])

organizations = pd.read_csv(RAW_DATA_PATH / "organizations.csv",
    dtype={"Id": str, "ZIP": str, "PHONE": str})

providers = pd.read_csv(RAW_DATA_PATH / "providers.csv",
    dtype={"Id": str, "ORGANIZATION": str, "ZIP": str})

payers = pd.read_csv(RAW_DATA_PATH / "payers.csv",
    dtype={"Id": str, "ZIP": str, "PHONE": str})

conditions = pd.read_csv(RAW_DATA_PATH / "conditions.csv",
    dtype={"PATIENT": str, "ENCOUNTER": str, "CODE": str},
    parse_dates=["START", "STOP"])
```

### Normalize Columns

```python
def normalize_columns(df: pd.DataFrame) -> pd.DataFrame:
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("-", "_")
    )
    return df

patients = normalize_columns(patients)
encounters = normalize_columns(encounters)
organizations = normalize_columns(organizations)
providers = normalize_columns(providers)
payers = normalize_columns(payers)
conditions = normalize_columns(conditions)
```

### Rename columns to match the expected naming convention

```python
# Rename the "id" columns to more descriptive names for clarity
patients.rename(columns={"id": "patient_id"}, inplace=True)
encounters.rename(columns={"id": "encounter_id"}, inplace=True)
organizations.rename(columns={"id": "organization_id"}, inplace=True)
providers.rename(columns={"id": "provider_id"}, inplace=True)
payers.rename(columns={"id": "payer_id"}, inplace=True)

# Rename the foreign key columns in the encounters DataFrame to match the new naming convention
encounters.rename(columns={"patient": "patient_id", "organization": "organization_id", "provider": "provider_id", "payer": "payer_id"}, inplace=True)

# Rename the foreign key columns in the conditions DataFrame to match the new naming convention
conditions.rename(columns={"patient": "patient_id", "encounter": "encounter_id"}, inplace=True)
```

### Referential Integrity Checks


```python

print("=" * 60)
print("REFERENTIAL INTEGRITY REPORT")
print("=" * 60)

# Get valid ID sets
valid_patient_ids = set(patients["patient_id"].dropna())
valid_org_ids = set(organizations["organization_id"].dropna())
valid_provider_ids = set(providers["provider_id"].dropna())
valid_payer_ids = set(payers["payer_id"].dropna())
valid_encounter_ids = set(encounters["encounter_id"].dropna())

# Check 1: Encounters → Patients
bad_patient = encounters[~encounters["patient_id"].isin(valid_patient_ids)]
print(f"\n Encounters referencing INVALID patient_id: {len(bad_patient)}")
if len(bad_patient):
    print(f" Example: {bad_patient['patient_id'].iloc[0]}")

# Check 2: Encounters → Organizations
bad_org = encounters[~encounters["organization_id"].isin(valid_org_ids)]
print(f"\n Encounters referencing INVALID organization_id: {len(bad_org)}")
if len(bad_org):
    print(f"Example: {bad_org['organization_id'].iloc[0]}")

# Check 3: Encounters → Providers
bad_provider = encounters[~encounters["provider_id"].isin(valid_provider_ids)]
print(f"\n Encounters referencing INVALID provider_id: {len(bad_provider)}")
if len(bad_provider):
    print(f"Example: {bad_provider['provider_id'].iloc[0]}")

# Check 4: Encounters → Payers
bad_payer = encounters[~encounters["payer_id"].isin(valid_payer_ids)]
print(f"\n Encounters referencing INVALID payer_id: {len(bad_payer)}")
if len(bad_payer):
    print(f"Example: {bad_payer['payer_id'].iloc[0]}")

# Check 5: Conditions → Encounters
bad_encounter = conditions[~conditions["encounter_id"].isin(valid_encounter_ids)]
print(f"\n Conditions referencing INVALID encounter_id: {len(bad_encounter)}")
if len(bad_encounter):
    print(f"Example: {bad_encounter['encounter_id'].iloc[0]}")

# Row counts summary
print("\n" + "=" * 60)
print(" ROW COUNTS SUMMARY")
print("=" * 60)
print(f"patients.csv      : {len(patients):>10,} rows")
print(f"encounters.csv    : {len(encounters):>10,} rows")
print(f"organizations.csv : {len(organizations):>10,} rows")
print(f"providers.csv     : {len(providers):>10,} rows")
print(f"payers.csv        : {len(payers):>10,} rows")
print(f"conditions.csv    : {len(conditions):>10,} rows")

# FINAL STATUS
all_bad = len(bad_patient) + len(bad_org) + len(bad_provider) + len(bad_payer) + len(bad_encounter)
print("\n" + "=" * 60)
if all_bad == 0:
    print("ALL CHECKS PASSED - Referential Integrity OK!")
else:
    print(f" FOUND {all_bad} BROKEN REFERENCES - Review before loading!")
print("=" * 60)
```
